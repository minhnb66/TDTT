```python
#b8
ten = input("Ten chu ho: ")
prev = int(input("Chi so thang truoc: "))
curr = int(input("Chi so thang nay: "))
diff = curr - prev
if 0 <= diff <= 50: noVAT = diff * 1984
elif 51 <= diff <= 100: noVAT = 50*1984 + (diff-50)*2050
elif 101 <= diff <= 200: noVAT = 50*1984 + 50*2050 + (diff-100)*2380
elif 201 <= diff <= 300: noVAT = 50*1984 + 50*2050 + 100*2380 + (diff-200)*2998
elif 301 <= diff <= 400: noVAT = 50*1984 + 50*2050 + 100*2380 + 100*2998 + (diff-300)*3350
else: noVAT = 50*1984 + 50*2050 + 100*2380 + 100*2998 + 100*3350 + (diff-400)*3460
total = noVAT * 1.08
print(f"Ho va ten: {ten}")
print(f"Tien phai tra la: {total:.0f}")
```
