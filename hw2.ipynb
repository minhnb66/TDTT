```python
#b3
a, b, c = map(int, input().split())
if a+b>c and a+c>b and b+c>a and a>0 and b>0 and c>0: 
    vuong = (a**2+b**2==c**2) or (a**2+c**2==b**2) or (b**2+c**2==a**2)
    can = a == b or b == c or c == a
    if a == b == c: print("Tam giác đều")
    elif vuong and can: print("Tam giác vuông cân")
    elif vuong: print("Tam giác vuông")
    elif can: print("Tam giác cân")
    else: print("Tam giác thường")
    hp = (a + b + c) / 2
    s = (hp * (hp-a) * (hp-b) * (hp-c)) ** 0.5
    p = hp * 2
    print(f"{s:.2f} {p:.2f}")
else: print("Không là ba cạnh của một tam giác")
```

```python
#b5
a, b, c = map(int, input().split())
if a == 0:
    if b == 0:
        if c == 0: print("Vô số nghiệm")
        else: print("Vô nghiệm")
    else: print(f"{-c/b:.2f}")
else: 
    delta = b**2 - 4*a*c
    if delta < 0: print("Vô nghiệm")
    elif delta == 0: print(f"{-b/(2*a):.2f}")
    elif delta > 0: 
        x1 = (-b + delta**0.5) / (2*a)
        x2 = (-b - delta**0.5) / (2*a)
        print(f"{x1:.2f} {x2:.2f}")
```

```python
#b8
a, b, c, d, m, n = map(int, input().split())
det = a*d - b*c
dx = m*d - b*n
dy = a*n - m*c
if det != 0:
    x = dx / det
    y = dy/det
    print(f"{x:.2f} {y:.2f}")
else:
    if dx == 0 and dy == 0: print("Vô số nghiệm")
    else: print("Vô nghiệm")
```
