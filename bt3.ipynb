```python
#A1
n = int(input())
if 100 <= n <= 999:
    a = n // 100
    b = (n // 10) % 10
    c = n % 10
    res = c*100 + b*10 + a
    print(res)
else: print("INVALID")
```

```python
#A16
n = float(input())
flr = n // 1
if flr == n: cl = flr
else: cl = flr + 1
if n - flr >= 0.5: rnd = flr + 1
else: rnd = flr
print(int(cl), int(flr), int(rnd))
```

```python
#A17
a, b, c, d = map(int, input().split())
if a != 0 and b % a == 0:
    q = b // a
    if b * q == c and c * q == d: print("YES")
    else: print("NO")
elif a==0 and b==0 and c==0 and d==0: print("YES")
else: print("NO")
```
